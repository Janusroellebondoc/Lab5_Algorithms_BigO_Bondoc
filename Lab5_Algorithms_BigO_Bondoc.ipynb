```python
# CELL 1: PYTHON RUNTIME TEST

import sys
import platform

print("Python Version:", sys.version.split()[0])
print("Operating System:", platform.system())
print("Python Executable:", sys.executable)
```

```python
# CELL 1: STUDENT CONFIGURATION

LAST_NAME = "Bondoc"
STUDENT_ID = "TUPM-26-1268"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
```

```python
# CELL 2: LARGEST-VALUE ALGORITHM

data = [
    10 + SEED_DIGIT,
    5 + NAME_LENGTH,
    20 + SEED_DIGIT,
    8 + NAME_LENGTH,
    15 + ID_SUM
]

largest = data[0]
comparisons = 0

for value in data[1:]:
    comparisons += 1
    if value > largest:
        largest = value

print("Data:", data)
print("Largest:", largest)
print("Comparisons:", comparisons)
```

```python
# CELL 3: LINEAR GROWTH

def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
# CELL 4: BIG O CLASSIFICATION

def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | "
        f"O(1)={1:4d} | "
        f"O(log n)={log_ops(n):4d} | "
        f"O(n)={n:4d} | "
        f"O(n^2)={n*n:7d}"
    )
```

```python
# CELL 5: SPACE COMPLEXITY

def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(
        f"n={n:5d} | "
        f"O(1) Space | "
        f"O(n) Space: {len(data)}"
    )
```

```python
# CELL 6: CASE ANALYSIS

def search_case(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i + 1
    return len(data)

data = [11, 22, 33, 44, 55]

for case, target in [
    ("Best", 11),
    ("Average", 33),
    ("Worst", 99)
]:
    print(f"{case} Case: {search_case(data, target)} comparisons")
```

```python
# CELL 7: LINEAR SEARCH

def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# CELL 8: BINARY SEARCH

def binary_search(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1, comparisons

index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# CELL 9: SEARCH COMPARISON

data = list(range(1, 101))
target = ((ID_SUM + SEED_DIGIT) % 100) + 1

li, lc = linear_search(data, target)
bi, bc = binary_search(data, target)

print("Target:", target)
print(f"Linear -> index={li}, comparisons={lc}")
print(f"Binary -> index={bi}, comparisons={bc}")
print("Reduction:", lc - bc)
```

```python
# CELL 10: BUBBLE SORT

def bubble_sort(data):
    a = data.copy()
    comparisons, swaps = 0, 0

    for i in range(len(a)):
        swapped = False

        for j in range(len(a) - i - 1):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True

        if not swapped:
            break

    return a, comparisons, swaps

data = [
    (ID_SUM * 3) % 100,
    (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100,
    42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# CELL 11: SELECTION SORT

def selection_sort(data):
    a = data.copy()
    comparisons, swaps = 0, 0

    for i in range(len(a) - 1):
        minimum = i

        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j

        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1

    return a, comparisons, swaps

sorted_data, comparisons, swaps = selection_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# CELL 12: INSERTION SORT

def insertion_sort(data):
    a = data.copy()
    comparisons, shifts = 0, 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons += 1
            if a[j] > key:
                a[j + 1] = a[j]
                shifts += 1
                j -= 1
            else:
                break

        a[j + 1] = key

    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
# CELL 13: EXECUTION-TIME MEASUREMENT

import time

def measure(sort_function, data, trials=5):
    times = []

    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        elapsed = time.perf_counter() - start
        times.append(elapsed)

    return result, sum(times) / trials

data_size = 100 + ID_SUM * 10

performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print("Dataset Size:", data_size)
print(f"Bubble Sort: {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
# CELL 14: SORTING PERFORMANCE COMPARISON

results = [
    ("Bubble Sort", bubble_result, bubble_time),
    ("Selection Sort", selection_result, selection_time),
    ("Insertion Sort", insertion_result, insertion_time)
]

print(
    f"{'Algorithm':<18}"
    f"{'Comparisons':>15}"
    f"{'Movements':>12}"
    f"{'Time (s)':>14}"
)

for name, result, elapsed in results:
    sorted_values, comparisons, movements = result

    print(
        f"{name:<18}"
        f"{comparisons:>15}"
        f"{movements:>12}"
        f"{elapsed:>14.8f}"
    )
```
